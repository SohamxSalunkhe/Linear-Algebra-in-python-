```python
import pandas as pd
```

```python
pa=pd.read_csv("/content/streamverse_activity (1).csv")
print("DATA is Here")
print(pa.head())
```

```python
print(pa.isnull().sum())
```

```python
gen_avg=pa.groupby("video_genre")["user_rating"].transform("mean")

pa["user_rating"]=pa["user_rating"].fillna(gen_avg)
```

```python
print(pa.isnull().sum())
```

```python
#2
pa["watch_date"] = pd.to_datetime(pa["watch_date"],)
print(pa["watch_date"].dtype)
```

```python
#3
pa["watch_time_hours"]=pa["watch_time_minutes"]/60

print(pa[["watch_time_minutes","watch_time_hours"]].head())
```

```python
print(pa.isnull().sum())
```

```python
# load
pa.to_csv("user_activity_warehouse.csv",index=False)
```

```python
#data mart
US=pa[(pa["country"]=="USA")&(pa["user_rating"]>=4.5)].copy()
```

```python
US.to_csv("top_rated_us_content_mart.csv", index=False)
print(US)
```
